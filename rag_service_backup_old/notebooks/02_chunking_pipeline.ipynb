```python
from app.ingestion.chunker import chunk_document
from app.ingestion.parsers import ParsedDocument, ParsedSection

doc = ParsedDocument(
    document_id='DOC_CHUNKING_TEST',
    file_name='distributed_systems.txt',
    full_text='',
    sections=[
        ParsedSection(heading='CAP Theorem', content='In distributed data stores, it is impossible to simultaneously provide Consistency, Availability, and Partition tolerance.', level=1),
        ParsedSection(heading='PACELC Theorem', content='In case of Partition, trade off Availability and Consistency; Else, trade off Latency and Consistency.', level=2)
    ]
)

chunks = chunk_document(doc, 'DOC_CHUNKING_TEST')
print(f'Produced {len(chunks)} structure-aware chunks.')
for i, c in enumerate(chunks):
    print(f'Chunk {i}: id={c.chunk_id}, len={len(c.text)}, section={c.metadata.get("section")}')
```
